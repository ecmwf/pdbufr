# Tropical cyclone reader: overview

```python
import pdbufr
```

#### The default settings

```python
df = pdbufr.read_bufr("tropical_cyclone.bufr", reader="tropical-cyclone", filters={"storm_id": "27W", "member": 1})
df.head(8)
```

#### Filters

```python
pdbufr.read_bufr(
    "tropical_cyclone.bufr",
    reader="tropical-cyclone",
    filters={"storm_id": "27W", "member": 1, "step": slice(0, 48)},
)
```

```python
pdbufr.read_bufr(
    "tropical_cyclone.bufr",
    reader="tropical-cyclone",
    filters={"count": 2, "member": [1, 2], "step": lambda s: s % 24 == 0 and s <= 48},
)
```

#### Selecting columns

```python
pdbufr.read_bufr(
    "tropical_cyclone.bufr",
    reader="tropical-cyclone",
    columns=["storm_id", "member", "step", "lat", "lon", "mslp"],
    filters={"storm_id": "27W", "member": 1, "step": slice(0, 24)},
)
```

```python
df = pdbufr.read_bufr(
    "tropical_cyclone.bufr",
    reader="tropical-cyclone",
    columns="all",
    filters={"storm_id": "27W", "member": 1, "step": slice(0, 24)},
)
df
```

#### Missing values

```python
filters = {"storm_id": "27W", "member": 1}
df_default = pdbufr.read_bufr("tropical_cyclone.bufr", reader="tropical-cyclone", filters=filters)
df_all = pdbufr.read_bufr("tropical_cyclone.bufr", reader="tropical-cyclone", filters=filters, keep_missing=True)
len(df_default), len(df_all)
```

```python
df_all.tail(3)
```

#### Units

```python
pdbufr.read_bufr(
    "tropical_cyclone.bufr",
    reader="tropical-cyclone",
    columns=["storm_id", "member", "step", "lat", "lon", "mslp"],
    filters={"storm_id": "27W", "member": 1, "step": slice(0, 24)},
    units={"mslp": "hPa"},
    units_columns=True,
)
```
