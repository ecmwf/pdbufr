# Flat reader: aircraft data

```python
import pdbufr
```

## Using columns

By default all the header and data keys are extracted.

```python
df = pdbufr.read_bufr("aircraft_small.bufr", reader="flat")
df.T
```

However, we can extract only the **header keys**:

```python
df = pdbufr.read_bufr("aircraft_small.bufr", columns="header", reader="flat")
df.T[:6]
```

or only the **data keys**:

```python
df = pdbufr.read_bufr("aircraft_small.bufr", columns="data", reader="flat")
df.T[:18]
```

## Filtering

Filtering works similarly to the hierarchical (i.e. non-flat) mode. In the following example only the messages/subsets matching a given flight number are extracted.

```python
df = pdbufr.read_bufr(
    "aircraft_small.bufr", columns="data", filters={"aircraftFlightNumber": "UOZDOZ2S"}, reader="flat"
)
df.T.iloc[:18]
```
