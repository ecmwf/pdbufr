```python
import pdbufr
```

## Periods

```python
pdbufr.read_bufr(
    "synop_150_85_M22.bufr",
    reader="synop",
    columns=["stnid", "max_wgust_speed"],
)
```

The period is named by its duration and the unit: "min" for minutes and "h" for hours. The periods are counted backwards from the observation time. Here the maximum and minimum temperature are for the last 12 hours:

```python
pdbufr.read_bufr(
    "synop_0_74_M32.bufr",
    reader="synop",
    columns=["stnid", "max_t2m", "min_t2m"],
)
```

## Values without a period

## Levels

```python
pdbufr.read_bufr(
    "syn_new.bufr",
    reader="synop",
    columns=["stnid", "t2m", "td2m"],
    level_columns=True,
)
```

## Several sensors

When several sensors report the same parameter for the same period, the value from the sensor closest to the nominal height is used (2 m for the maximum/minimum 2 m temperature) and it keeps its own level. Sensors below 0.5 m (e.g. a grass sensor) are ignored. The following message contains the 60 minute minimum temperature measured at 2 m (275.45 K) and at 0.05 m (274.95 K). The 2 m value is selected:

```python
pdbufr.read_bufr(
    "synop_172_98_M29.bufr",
    reader="synop",
    columns=["stnid", "min_t2m"],
    level_columns=True,
)
```
