# Synop reader: cloud columns

```python
import pdbufr
```

#### The cloud types

```python
df = pdbufr.read_bufr(
    "syn_new.bufr",
    reader="synop",
    columns=["station", "cloud_cover", "cloud_type_low", "cloud_type_middle", "cloud_type_high"],
)
```

```python
df.T
```

#### The cloud amount and the cloud base

```python
df = pdbufr.read_bufr(
    "syn_new.bufr",
    reader="synop",
    columns=[
        "station",
        "cloud_cover",
        "cloud_type_low",
        "cloud_type_middle",
        "cloud_type_high",
        "cloud_amount_nh",
        "cloud_base_lowest",
    ],
)
```

```python
df.T
```

#### Filtering on a cloud column

```python
df = pdbufr.read_bufr(
    "syn_new.bufr",
    reader="synop",
    columns=["station", "cloud_amount_nh"],
    filters={"cloud_amount_nh": 2},
)
```

```python
df.T
```

#### Units and unit conversion

```python
df = pdbufr.read_bufr(
    "syn_new.bufr",
    reader="synop",
    columns=["station", "cloud_cover", "cloud_type_low", "cloud_amount_nh", "cloud_base_lowest"],
    units_columns=True,
)
```

```python
df.T
```

#### Converting the cloud base

```python
df = pdbufr.read_bufr(
    "syn_new.bufr",
    reader="synop",
    columns=["station", "cloud_base_lowest", "cloud_amount_nh"],
    units_columns=True,
    units={"cloud_base_lowest": "ft"},
)
```

```python
df.T
```
